```python

# graph (Dictionary of Dictionaries)

graph = {
    'Start': {'A': 5, 'B': 2},
    'A': {'C': 4, 'Goal': 7},
    'B': {'A': 1, 'D': 3},
    'C': {'Goal': 2},
    'D': {'Goal': 8},
    'Goal': {}
}

# start and goal (Strings)

start_node = 'Start'
goal_node = 'Goal'

# heuristic (Dictionary)

heuristic = {
    'Start': 8,
    'A': 5,
    'B': 6,
    'C': 2,
    'D': 7,
    'Goal': 0
}

# frontier (List of Tuples)
# What it does: The "waiting list" of nodes to explore next. 
# Each item is a tuple containing (node_name, priority_score).
# *Note: In real-world apps, use Python's `heapq` module instead of a list.
frontier = [('C', 4), ('B', 7), ('D', 15)]

# visited (Set)
# What it does: Keeps track of nodes that have already been completely explored.
# Why a Set?: Optimized for "membership testing" (O(1) lookup time) 
# to prevent infinite loops if your graph has cycles.
visited = {'Start', 'A'}

# cost_so_far (Dictionary)
# What it does: Used in UCS and A*. Represents g(n). 
cost_so_far = {
    'Start': 0,
    'A': 5,
    'B': 2
}


# ---------------------------------------------------------------------
# 3. THE BREADCRUMB TRAIL (Reconstructing the Answer)
# ---------------------------------------------------------------------

# came_from (Dictionary)
# What it does: Maps a node to its "parent" (the node that immediately 
# preceded it on the best path). When the algorithm hits the goal, 
# it loops backward through this dictionary until it hits the start node.
came_from = {
    'Start': None,
    'A': 'Start',
    'B': 'Start',
    'C': 'A',
    'Goal': 'C'
}

# path (List)
final_path = ['Start', 'A', 'C', 'Goal']
```

```python
def greedy_bfs(graph, start, goal, heuristic):
    # 1. INITIALIZATION: Priority is strictly h(n)
    frontier = [(start, heuristic[start])]
    visited = set()
    came_from = {start: None}
    # Note: cost_so_far is completely ignored in Greedy BFS
    
    while frontier:
        frontier.sort(key=lambda x: x[1])
        current_node, _ = frontier.pop(0)
        
        if current_node in visited:
            continue
        visited.add(current_node)
        
        if current_node == goal:
            path = []
            while current_node is not None:
                path.append(current_node)
                current_node = came_from[current_node]
            path.reverse()
            print(f"Goal found with Greedy BFS. Path: {path}")
            return path
            
        for neighbor, edge_cost in graph.get(current_node, {}).items():
            # Note: We don't calculate new_cost since we don't track g(n)
            
            if neighbor not in came_from:  # Simple check to avoid infinite loops
                came_from[neighbor] = current_node
                
                # 2. THE FORMULA: priority = h(n)
                priority = heuristic[neighbor]
                frontier.append((neighbor, priority))
                
    print("Goal not found")
    return None
```

```python
def ucs(graph, start, goal):
    # 1. INITIALIZATION: Priority is strictly g(n)
    frontier = [(start, 0)]
    visited = set()
    came_from = {start: None}
    cost_so_far = {start: 0} # Tracks g(n)
    
    while frontier:
        frontier.sort(key=lambda x: x[1])
        current_node, _ = frontier.pop(0)
        
        if current_node in visited:
            continue
        visited.add(current_node)
        
        if current_node == goal:
            path = []
            while current_node is not None:
                path.append(current_node)
                current_node = came_from[current_node]
            path.reverse()
            print(f"Goal found with UCS. Path: {path}, Total Cost: {cost_so_far[goal]}")
            return path
            
        for neighbor, edge_cost in graph.get(current_node, {}).items():
            new_cost = cost_so_far[current_node] + edge_cost
            
            # Check if neighbor is unvisited OR if we found a shorter path to it
            if neighbor not in cost_so_far or new_cost < cost_so_far[neighbor]:
                cost_so_far[neighbor] = new_cost
                came_from[neighbor] = current_node
                
                # 2. THE FORMULA: priority = g(n)
                priority = new_cost
                frontier.append((neighbor, priority))
                
    print("Goal not found")
    return None
```

```python
def a_star_search(graph, start, goal, heuristic):
    # 1. INITIALIZATION: Priority is g(n) + h(n)
    frontier = [(start, 0 + heuristic[start])]
    visited = set()
    came_from = {start: None}
    cost_so_far = {start: 0} # Tracks g(n)
    
    while frontier:
        frontier.sort(key=lambda x: x[1])
        current_node, _ = frontier.pop(0)
        
        if current_node in visited:
            continue
        visited.add(current_node)
        
        if current_node == goal:
            path = []
            while current_node is not None:
                path.append(current_node)
                current_node = came_from[current_node]
            path.reverse()
            print(f"Goal found with A*. Path: {path}, Total Cost: {cost_so_far[goal]}")
            return path
            
        for neighbor, edge_cost in graph.get(current_node, {}).items():
            new_cost = cost_so_far[current_node] + edge_cost
            
            # Check if neighbor is unvisited OR if we found a shorter path to it
            if neighbor not in cost_so_far or new_cost < cost_so_far[neighbor]:
                cost_so_far[neighbor] = new_cost
                came_from[neighbor] = current_node
                
                # 2. THE FORMULA: priority = g(n) + h(n)
                priority = new_cost + heuristic[neighbor]
                frontier.append((neighbor, priority))
                
    print("Goal not found")
    return None
```
